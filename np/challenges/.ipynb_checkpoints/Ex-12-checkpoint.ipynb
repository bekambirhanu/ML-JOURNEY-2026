### E-commerce – Inventory Restocking
You have an array current_stock of shape (500,000,). Create a new array order_quantity that is 100 where current_stock is less than 50, and 0 otherwise, using np.where().

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

current_stock = rng.integers(20, 100, (500_000,))

order_quality = np.where(current_stock<50,100,0)
order_quality
```
