### Healthcare – Age Grouping
An array ages of shape (1,100,000,) contains patient ages. Create a Boolean array is_senior that is True for ages 65 and above.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)
ages = rng.integers(4, 80, (1, 100_000))

is_senior = np.where(ages>=65, True, False)
is_senior
```
