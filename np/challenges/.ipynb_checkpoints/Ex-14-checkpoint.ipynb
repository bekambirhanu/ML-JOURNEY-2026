### NumPy Basics – Element-wise Comparison
Two arrays actual_sales and forecast_sales, each of shape (2,500,000,), exist. Determine how many forecasts were too high (forecast > actual).

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

actual_sales = rng.integers(20, 100, (2,500_000))
forecast_sales = rng.integers(30, 110, (2,500_000))

condition = actual_sales>forecast_sales
actual_sales.sum(where=condition,axis=0).sum()
```
