### Reshaping – Data Alignment
You have a 1D array of 1,000,000 sensor readings. Reshape it into a 2D array with 1,000 rows and 1,000 columns to prepare it for heatmap visualization.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

sensor_readings = rng.integers(10, 100, (1_000_000))
reshaped = sensor_readings.reshape((1000,1000))
reshaped
```
