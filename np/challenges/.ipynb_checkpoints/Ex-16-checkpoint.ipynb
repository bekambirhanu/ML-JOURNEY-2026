### E-commerce – Tiered Discounts
A purchase_amounts array (2,000,000,) exists. Apply a discount using np.select(): 5% off for purchases $50-$99.99, 10% off for $100-$199.99, and 15% off for $200 and above. No discount below $50.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

purchase_amounts = np.abs(rng.standard_normal((2_000_000,))) *100

condition_list = [(purchase_amounts>=50)&(purchase_amounts<=99.99), (purchase_amounts>=100)&(purchase_amounts<=199.99), purchase_amounts>=200]
choice_list = [purchase_amounts-purchase_amounts*0.05, purchase_amounts-purchase_amounts*0.1, purchase_amounts-purchase_amounts*0.15]

discount = np.select(condlist=condition_list, choicelist=choice_list)
```
