### Finance – Portfolio Volatility
You have a NumPy array daily_returns of shape (1,000,000, 10) representing daily returns for 10 assets over 1 million days. Calculate the annualized volatility (standard deviation * sqrt(252)->trading days in a year) for each asset using a single vectorized operation.

```python
import numpy as np

rng = np.random.default_rng(12)
np.set_printoptions(precision=4, suppress=True)
```

```python
# the daily returns of 10 stocks from different companies
daily_returns = rng.standard_normal(size= (1_000_000, 10))

# the number of trading days annualy
trading_days = 252

# The standard deviation of the 10 companies
std = np.std(a=daily_returns, axis=0)

# Therefore, the annual volatility from a million daily returns
volatility = std * np.sqrt(trading_days)
volatility
```
