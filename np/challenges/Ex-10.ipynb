### Finance – Simple Moving Average
You have a 1D array stock_prices of shape (1,000,000,). Calculate the 50-day simple moving average (SMA) using `np.convolve`.

```python
import numpy as np
rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
```

```python
stock_prices = rng.standard_normal(size=(1_000_000))*100

days = np.ones((50,))/50

sma_of_50_day = np.convolve(stock_prices, days, "valid")
sma_of_50_day
```
