### Environmental Science – Pressure Normalization
You have pressure_readings of shape (1,200,000,). Normalize the data to a range of 0 to 1 using Min-Max scaling: (x - min) / (max - min).

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(seed=1234)
pressure_readings = rng.integers(101_000,189_000,size=(1, 200_000))
normalized_pressure = (pressure_readings-pressure_readings.min())/(pressure_readings.max()-pressure_readings.min())
print(normalized_pressure)
```
