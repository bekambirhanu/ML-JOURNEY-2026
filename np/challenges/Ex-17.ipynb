### Healthcare – Risk Categorization
An array bmi_values (1,500,000,). Using np.select(), categorize patients into risk categories: "Underweight" (<18.5), "Normal" (18.5-24.9), "Overweight" (25-29.9), and "Obese" (>=30).

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

bmi_values = rng.normal(23, 4.5, size=(1_500_000))

conditions = [bmi_values<18.5, (bmi_values>=18.5)&(bmi_values<=24.9), (bmi_values>=24.9)&(bmi_values<=29.9), bmi_values>=29.9]
choices = ["Underweight", "Normal", "Overweight", "Obese"]

categorised = np.select(conditions, choices, default="NaN")
print(bmi_values)
print(categorised)
```
