### Finance – Portfolio Rebalancing
You have a portfolio weights array (10, 1,000,000) representing the weight of 10 assets in 1,000,000 portfolios. The sum of weights per portfolio must be 1. Normalize each portfolio so the weights sum to 1 using broadcasting and np.sum().

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

portfolio = rng.integers(10, 1_000_000, (10, 1_000_000))
totals_of_assets = portfolio.sum(axis=0)
normalized = (portfolio[:,]/totals_of_assets)
normalized.sum(axis=0).round(decimals=0)
```
