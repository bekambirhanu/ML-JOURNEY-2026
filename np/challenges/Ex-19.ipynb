### Environmental Science – Sensor Data Correction
You have sensor_data (5,000,000,) and a calibration_offset (5,000,000,) that varies per reading. Correct the data by adding the offset. Then, apply a drift_correction factor (scalar) using broadcasting.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

sensor_data = rng.integers(21, 41, (5_000_000,))
offsets = rng.normal(0, 2, (5_000_000,))
drift_correction = 0.8

correct = sensor_data - offsets
final_corrected_data = correct*drift_correction

final_corrected_data
```
