### Healthcare – BMI Calculation
A hospital has heights (in meters) and weights (in kg) arrays, each of shape (1,200,000,). Calculate the Body Mass Index (BMI) for all patients simultaneously.



```python
import numpy as np

rng = np.random.default_rng(seed=[123])
np.set_printoptions(precision=6, suppress=True)
```

```python
heights = rng.normal(loc=1.78,size=(1, 200_000), scale=0.07)
weights = rng.normal(loc=65, size=(1, 200_000), scale=15)

bmi = weights/heights**2
bmi
```
