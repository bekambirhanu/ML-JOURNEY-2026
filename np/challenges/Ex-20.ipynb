### E-commerce – Customer Lifetime Value (CLV)
Arrays avg_purchase_value (500,000,) and purchase_frequency (500,000,) exist. Calculate CLV using the formula: avg_purchase_value * purchase_frequency * 2.5, vectorized.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

avg_purchase_value = rng.integers(20, 3999, (500_000))
purchase_frequency = rng.integers(1, 11, (500_000))

clv = avg_purchase_value*purchase_frequency*2.5
clv
```
