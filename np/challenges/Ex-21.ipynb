### Image Processing – Image Brightening
An image array of shape (4000, 3000, 3) (uint8) is too dark. Add a scalar value of 40 to all pixels, then clip the result to the valid range [0, 255] to prevent overflow.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

image = rng.integers(0, 256, (4000, 3000, 3), dtype=np.uint8)
scaled_up = image + 40
brightened_image = scaled_up.clip(max= 255)
brightened_image
```
