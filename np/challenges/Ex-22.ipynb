### Data Analysis – Conditional Aggregation
You have sales_data (1,000,000, 2), where column 0 is "region_code" (0-9) and column 1 is "sales_amount". Calculate the total sales for region 5 without using a loop or explicit indexing beyond a Boolean mask.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)
region_code = rng.integers(0, 10, (1_000_000,)); sales_amount = rng.integers(10_000, 1_500_001, (1_000_000,))
sales_data = np.zeros((1_000_000, 2))
sales_data[:, 0] = region_code; sales_data[:, 1] = sales_amount

total_sales_of_r5 = sales_data[sales_data[:, 0] == 5][:, 1].sum()
total_sales_of_r5
```
