### Finance – Moving Average Crossover
You have prices (1,000,000,). Calculate the 50-day and 200-day moving averages. Create a Boolean array signal where True indicates a "golden cross" (50-day > 200-day) and False otherwise.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

price = rng.integers(10_000, 1_500_000,(1_000_000,))
day_50 = np.ones((50)); day_200 = np.ones((200))

ma_50 = np.convolve(price, day_50, "valid")
ma_200 = np.convolve(price, day_200, "valid")

is_golden_cross = ma_50[:len(ma_200)] > ma_200

is_golden_cross
```
