### Environmental Science – Wind Chill Calculation
You have temp_f (2,000,000,) and wind_speed_mph (2,000,000,). Calculate the wind chill using the formula: 35.74 + 0.6215*T - 35.75*(V**0.16) + 0.4275*T*(V**0.16).

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

temp_f = rng.integers(-3, 50, (2_000_000,))
wind_speed = rng.normal(1456, 200, (2_000_000))

wind_chill = 35.74 + 0.6215*temp_f - 35.75*(wind_speed**0.16) + 0.4275*temp_f*(wind_speed**0.16)
wind_chill
```
