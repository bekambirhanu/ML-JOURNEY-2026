### Advanced Indexing – Replacing Values
A pixel_array (5000, 5000) contains values 0-255. Using advanced indexing, replace all pixels with a value less than 50 with 0, and all pixels greater than 200 with 255 in a single step.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

pixel_array = rng.integers(0, 256, (5000, 5000))

conditions = [pixel_array<50, pixel_array>200]
choices = [0, 255]

result_pixles = np.select(conditions, choices, default=pixel_array)
result_pixles
```
