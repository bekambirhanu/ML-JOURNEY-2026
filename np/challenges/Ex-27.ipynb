### E-commerce – Purchase Cohort Analysis
You have a 2D array cohort_data (100,000, 12) representing monthly purchases for 100,000 customers over 12 months. Calculate the cumulative sum for each customer across months using np.cumsum() with the correct axis.

```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

```python
rng = np.random.default_rng(1234)

cohort_data = rng.integers(299, 100_000, (100_000, 12))

cum_sum = cohort_data.cumsum(axis=1)
cum_sum
```
