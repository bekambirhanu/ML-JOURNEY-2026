### E-commerce – Tax Calculation
You have an array prices of shape (2,000,000,) representing the price of items sold. Apply a 7.5% sales tax to all items using vectorized multiplication.

```python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
```

```python
price_index = rng.random(size=2_000_000, dtype=np.float32)*60

sales_tax = 0.075

total_cost = (price_index + price_index*sales_tax).round(decimals=2)
total_cost
```
