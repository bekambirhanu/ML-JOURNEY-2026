### Environmental Science – Unit Conversion
A sensor network outputs temperature in Celsius in an array temp_c of shape (5,000,000,). Convert the entire array to Fahrenheit using the formula (C * 9/5) + 32


```python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
```

```python
temp_c = rng.random(size=5_000_000)*100

def conv_f(c: np.ndarray):
    return c*(9/5) + 32

temp_f = conv_f(temp_c)
temp_f
```
