### Finance – Currency Conversion
You have an array usd_balances of shape (1,500,000,) representing account balances in USD. Convert these balances to Euros using an exchange rate of 0.92, applied in a single operation.

```python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
```

```python
usd_balances = rng.integers(36, 100, size=1_500_000)

rate = 0.92

balance_in_eur = usd_balances*rate
balance_in_eur
```
