### Healthcare – Heart Rate Threshold
An array heart_rates of shape (2,000,000,) contains patient BPM. Create a Boolean array that is True where the heart rate is outside the normal range of 60-100 BPM.

```python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
```

```python
heart_rate = rng.integers(45, 200, 2_000_000)

# For the rates under the normal range
heart_rate[(heart_rate>=60) & (heart_rate<=100)] = True

# For the rates above or below the normal range
heart_rate[((heart_rate<60) & (heart_rate != 1)) | (heart_rate>100)] = False

normal_hr = np.bool(heart_rate)
normal_hr
```
