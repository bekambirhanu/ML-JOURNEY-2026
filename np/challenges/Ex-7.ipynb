### E-commerce – Discount Application
An array cart_values of shape (1,800,000,) holds the total cart value. Apply a 10% discount to all values greater than $100 using vectorized multiplication with a Boolean mask.

```python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
```

```python
# Generate the cart_values
cart_values = np.abs(rng.standard_normal(size=(1, 800_000)) *100)
print(cart_values)

# Reduce the vales that are over 100 by 10%
cart_values[0][cart_values[0]>100] -= 0.1*cart_values[0][cart_values[0]>100]

cart_values
```
