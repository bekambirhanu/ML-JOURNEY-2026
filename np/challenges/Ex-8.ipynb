### Image Processing – Grayscale Conversion
You have an image represented as a NumPy array rgb_image of shape (2000, 2000, 3). Convert it to grayscale using the formula:
```0.2989 * R + 0.5870 * G + 0.1140 * B```

````python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
````

````python
# Generate the rgb image rep
rgb_image = rng.integers(0, 256, size=(2000, 2000, 3)).astype(np.float64)
print(rgb_image)

# Convert it to a greyscale rep by using the given formula
rgb_image[:,:,0] *= 0.2989
rgb_image[:,:,1] *= 0.5870
rgb_image[:,:,2] *= 0.1140

grey_scale = np.sum(rgb_image,axis=2)
grey_scale
````
