### Data Sanitization – Outlier Capping
A temperature array of shape (3,000,000,) has erroneous readings below -50 and above 100. Cap these values to -50 and 100 respectively using ```np.clip()```.
cap- to identify those erroneous elements and to cut them to the maximum or minimum temperature.

````python
import numpy as np

rng = np.random.default_rng(seed=1234)
np.set_printoptions(precision=4, suppress=True)
````

````python
temperature_read = (rng.standard_normal(size=(3_000_000))*100).astype(np.int64)
print(temperature_read)
capped_temperature_read = np.clip(a=temperature_read, a_min=-50, a_max=100)
capped_temperature_read
````
