```python
# Question NO 10
np.random.seed(42)  # for reproducibility
n_houses = 50
living_area = np.random.normal(2000, 500, n_houses)
overall_qual = np.random.randint(1, 11, n_houses)  # 1-10 rating
year_built = np.random.randint(1950, 2025, n_houses)
sale_price = (living_area * 150) + (overall_qual * 20000) + (year_built - 1900)*1000 + np.random.normal(0, 30000, n_houses)

# rounding the sales price
np.round(sale_price)
# a boolean mask
built_after_2000 = np.zeros_like(year_built, dtype=np.bool)
built_after_2000[year_built>=2000] = True
# Compute average sale price for houses with overall_qual >= 8
sale_price[overall_qual>=8].mean()
# Find the correlation between living_area and sale_price (np.corrcoef)
np.corrcoef(sale_price, living_area)
```

```python
# Last question (Optional)
# define an iterative calculation
def vector_calc(array):
    return sum([i**50 for i in array])

np.random.seed(1)
array = np.arange(1000, 1000000, dtype=np.float64)
array

# Python-calculation duration = 0.10572495400265325

start = time.perf_counter()
vector_calc(array=array)
end = time.perf_counter()
end-start
# NumPy calculation duration = 0.012605597003130242

start = time.perf_counter()
np.sum(array**50)
end = time.perf_counter()
end-start
```
