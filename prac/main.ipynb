### A Simple script to calculate the simple moving average of price

```python
import pandas as pd
import numpy as np
```

```python
# function that returns the SMA
def calculate_50_day_sma(array: list):
    _50_day = np.ones((50))

    _50_day_sma = np.convolve(array, _50_day/50, mode="valid")

    return _50_day_sma
```

```python
url = "https://docs.google.com/spreadsheets/d/1hrm30Na_oLKSgElGqFb9imj-gHdoYe85IeUObcZfEkg/export?format=csv"

aapl_price = pd.read_csv(url)

sma_price_50 = calculate_50_day_sma(aapl_price["Close"])
```
