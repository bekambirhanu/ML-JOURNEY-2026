## Numpy walk steps data

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(seed=123456)

nwalks = 5000
nsteps = 1000
draws = rng.integers(0, 2, size=(nwalks, nsteps)) # 0 or 1
steps = np.where(draws > 0, 1, -1)
walks = steps.cumsum(axis=1)


hit30 = (np.abs(walks) >= 30).any(axis=1)
crossing_times = (np.abs(walks[hit30])).argmax(axis=1) # Indexes of every walk where it crossed 30 negatively or positively
crossing_times.mean()

plt.plot(walks[1, :3])
```
